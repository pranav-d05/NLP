# WordNet Semantic Relationships

Identify **synonyms, antonyms, and hypernyms** using NLTK WordNet.

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.corpus import wordnet as wn
```

```python
def get_synonyms(word):
    synonyms = set()
    for syn in wn.synsets(word):
        for lemma in syn.lemmas():
            synonyms.add(lemma.name())
    return list(synonyms)

def get_antonyms(word):
    antonyms = set()
    for syn in wn.synsets(word):
        for lemma in syn.lemmas():
            if lemma.antonyms():
                antonyms.add(lemma.antonyms()[0].name())
    return list(antonyms)

def get_hypernyms(word):
    hypernyms = set()
    for syn in wn.synsets(word):
        for h in syn.hypernyms():
            for lemma in h.lemmas():
                hypernyms.add(lemma.name())
    return list(hypernyms)
```

```python
word = 'good'

print('Synonyms:', get_synonyms(word))
print('Antonyms:', get_antonyms(word))
print('Hypernyms:', get_hypernyms(word))
```

```python
sentence = "The movie was good and the acting was excellent"
words = sentence.split()

for w in words:
    print(f"\nWord: {w}")
    print("Synonyms:", get_synonyms(w)[:5])
    print("Antonyms:", get_antonyms(w))
    print("Hypernyms:", get_hypernyms(w)[:5])
```
